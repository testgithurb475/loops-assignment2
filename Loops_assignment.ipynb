# Python If-Else Practice Questions

## Section A: Discounts and Pricing


### 1. Check Discount Eligibility
> Write a program to check if a customer is eligible for a discount. If the total purchase is more than $100, apply a 10% discount and display the final price. Otherwise, display the total price as it is.


```python
# Question 1: Check Discount Eligibility
total_purchase = 120

if total_purchase > 100:
    final_price = total_purchase * 0.90  # Apply 10% discount
    print(f"Eligible for discount! Final price: ${final_price:.2f}")
else:
    print(f"No discount. Total price: ${total_purchase:.2f}")
```

### 2. Calculate Bulk Discount
> If a customer buys more than 5 items, apply a 15% discount on the total price. Otherwise, no discount is applied. Display the total price.


```python
# Question 2: Calculate Bulk Discount
items_bought = 6
total_price = 200.0

if items_bought > 5:
    total_price = total_price * 0.85  # Apply 15% discount
    print(f"Bulk discount applied! Total price: ${total_price:.2f}")
else:
    print(f"Total price: ${total_price:.2f}")
```

### 3. Membership Discount
> Check if the customer is a member (`is_member = True`). Members get a 20% discount; non-members get a 5% discount. Calculate and print the discounted price.


```python
# Question 3: Membership Discount
is_member = True
total_price = 150.0

if is_member:
    discounted_price = total_price * 0.80  # 20% discount for members
    print(f"Member discounted price: ${discounted_price:.2f}")
else:
    discounted_price = total_price * 0.95  # 5% discount for non-members
    print(f"Non-member discounted price: ${discounted_price:.2f}")
```

### 4. Seasonal Sale
> If today is a holiday (`is_holiday = True`), apply a 25% discount; otherwise, apply a 10% discount. Calculate the price after discount.


```python
# Question 4: Seasonal Sale
is_holiday = True
price = 80.0

if is_holiday:
    price_after_discount = price * 0.75  # 25% discount
    print(f"Holiday sale price: ${price_after_discount:.2f}")
else:
    price_after_discount = price * 0.90  # 10% discount
    print(f"Regular sale price: ${price_after_discount:.2f}")
```

### 5. Buy-One-Get-One-Free
> If a customer buys an even number of items, they get half of them for free. Otherwise, they pay for all. Calculate the number of items the customer has to pay for.


```python
# Question 5: Buy-One-Get-One-Free
total_items = 4

if total_items % 2 == 0:
    items_to_pay = total_items // 2
    print(f"Even items! Items to pay for: {items_to_pay}")
else:
    items_to_pay = total_items
    print(f"Odd items! Items to pay for: {items_to_pay}")
```

## Section B: Tax Calculations


### 6. Sales Tax
> If the price of an item is greater than $500, apply a luxury tax of 15%. Otherwise, apply a standard tax of 8%. Display the total price after tax.


```python
# Question 6: Sales Tax
item_price = 600.0

if item_price > 500:
    total_price = item_price * 1.15  # 15% luxury tax
    print(f"Luxury tax applied. Total price after tax: ${total_price:.2f}")
else:
    total_price = item_price * 1.08  # 8% standard tax
    print(f"Standard tax applied. Total price after tax: ${total_price:.2f}")
```

### 7. Income Tax

```python
# Question 7: Income Tax
annual_income = 60000.0

if annual_income > 50000:
    tax_amount = annual_income * 0.20
    print(f"Tax amount (20%): ${tax_amount:.2f}")
else:
    tax_amount = annual_income * 0.10
    print(f"Tax amount (10%): ${tax_amount:.2f}")
```

### 9. VAT Calculation
> If the item is marked as essential (`is_essential = True`), apply a VAT of 5%. Otherwise, apply a VAT of 12%. Display the final price.


```python
# Question 9: VAT Calculation
is_essential = True
price = 100.0

if is_essential:
    final_price = price * 1.05  # 5% VAT
    print(f"Essential item final price (5% VAT): ${final_price:.2f}")
else:
    final_price = price * 1.12  # 12% VAT
    print(f"Non-essential item final price (12% VAT): ${final_price:.2f}")
```

### 11. Free Shipping

### 12. Discount Code
> If a customer enters the correct discount code (`DISCOUNT10`), apply a 10% discount. Otherwise, charge the full amount.


```python
# Question 12: Discount Code
discount_code = "DISCOUNT10"
amount = 100.0

if discount_code == "DISCOUNT10":
    final_amount = amount * 0.90
    print(f"Discount applied! Final amount: ${final_amount:.2f}")
else:
    final_amount = amount
    print(f"Invalid code. Full amount: ${final_amount:.2f}")
```

### 13. Tiered Discounts
> Apply discounts based on the total price:
> - $0-$50: No discount.
> - $50-$100: 10% discount.
> - Over $100: 20% discount.


```python
# Question 13: Tiered Discounts
total_price = 75.0

if total_price <= 50:
    final_price = total_price
    print(f"No discount. Final price: ${final_price:.2f}")
elif total_price <= 100:
    final_price = total_price * 0.90  # 10% discount
    print(f"10% discount applied. Final price: ${final_price:.2f}")
else:
    final_price = total_price * 0.80  # 20% discount
    print(f"20% discount applied. Final price: ${final_price:.2f}")
```
